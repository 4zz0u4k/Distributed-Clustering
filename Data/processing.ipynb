```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
```

```python
data = pd.read_csv('marketing_campaign.csv', sep='\t')
data.head()
```

```python
print(f"Shape before Nan drop {data.shape[0]}")
data = data.dropna()
print(f"Shape after Nan drop {data.shape[0]}")
```

```python
response = data['Response']
response.to_csv('response.csv', index=False)
```

```python
data.drop(columns=['ID', 'Dt_Customer', 'Z_CostContact', 'Z_Revenue','Response'], inplace=True)
```

```python
data_encoded = pd.get_dummies(data, columns=['Education', 'Marital_Status'], drop_first=True)
```

```python
scaler = StandardScaler()
data = pd.DataFrame(scaler.fit_transform(data_encoded), columns=data_encoded.columns)
```
